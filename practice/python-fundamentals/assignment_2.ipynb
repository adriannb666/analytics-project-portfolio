# Question 1: Prime Factorization (2 Points)
## Write a function that takes an integer n and returns a dictionary where the keys are the prime factors and the values are their exponents in the factorization.
## Your function must pass all test cases


```python
def pf(n):
    factors = {}
    divisor = 2
    while divisor * divisor <= n:
        while n % divisor == 0:
            if divisor in factors:
                factors[divisor] += 1
            else:
                factors[divisor] = 1
            n = n // divisor
        divisor += 1
    if n > 1:
        factors[n] = 1
    return factors
```

```python
pf(25936724)
```

# Question 2: Custom Sorting (2 Points)
## Write a function that sorts a list of strings based on the number of vowels each string contains. If two words have the same number of vowels, they should be sorted alphabetically.
## Your function must pass all test cases.


```python
def count_vowels(word):
    count = 0
    for letter in word.lower():
        if letter in "aeiou":
            count += 1
    return count


def customsort(words):
    # Sort alphabetically first, then by vowel count.
    alphabetical = sorted(words)
    return sorted(alphabetical, key=count_vowels)
```

```python
customsort(["apple", "banana", "cherry", "grape", "mango", "orange"])
```

```python
customsort(["tiger", "elephant", "giraffe", "dolphin", "penguin", "koala"])
```

```python
customsort(["Canada", "Brazil", "Germany", "India", "Japan", "Mexico"])
```

# Question 3: Pascal's Triangle (3 Points)
## Write a function that generates the nth row of Pascal’s Triangle.
# Then display the first 19 rows of Pascal's Triangle using the function you defined.


```python
from math import factorial


def pascal_row(n):
    row = []
    for k in range(n):
        # Use the combination C(n - 1, k).
        value = factorial(n - 1) // (factorial(k) * factorial(n - 1 - k))
        row.append(value)
    return row


# Display the first 19 rows.
for n in range(1, 20):
    print(*pascal_row(n))
```
